```python
Q1=df["Fare"].quantile(0.25)
Q3=df["Fare"].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

df=df[(df["Fare"]>=lower)&(df["Fare"]<=upper)]
```

# Project 2: Titanic Data Cleaning & Passenger Analysis

## Objective

The objective of this project is to clean the Titanic dataset and analyze passenger survival trends using Python. This project includes data cleaning, handling missing values, removing duplicates, detecting outliers, performing statistical analysis, creating visualizations, and drawing meaningful insights through Exploratory Data Analysis (EDA).

## Dataset Description

The Titanic dataset contains information about passengers who traveled on the Titanic. It includes details such as passenger class, gender, age, fare, embarkation port, cabin, and survival status. This dataset is widely used for learning data analysis and machine learning.

## Steps Performed

1. Imported the required Python libraries.
2. Loaded the Titanic dataset using Pandas.
3. Explored the dataset using head(), info(), describe(), and shape().
4. Checked and handled missing values.
5. Removed duplicate records.
6. Detected and treated outliers using the IQR method.
7. Filtered passengers based on age, gender, and passenger class.
8. Grouped the data using groupby() to analyze survival rates.
9. Calculated statistical measures such as mean, median, variance, and standard deviation.
10. Created visualizations including bar chart, line chart, histogram, scatter plot, and heatmap.
11. Summarized the findings using Exploratory Data Analysis (EDA).

## Observations

- Female passengers had a higher survival rate than male passengers.
- First-class passengers had the highest survival rate.
- Most passengers were between 20 and 40 years of age.
- The Fare column contained some outliers.
- Passenger class and fare influenced survival.

## Conclusion

This project successfully cleaned and analyzed the Titanic dataset using Python, Pandas, NumPy, Matplotlib, and Seaborn. The analysis showed that gender, passenger class, and fare were important factors affecting passenger survival. Data visualization made it easier to understand the patterns in the dataset.
