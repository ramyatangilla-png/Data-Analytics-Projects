```python
plt.figure(figsize=(8,5))
plt.hist(df['Sales'], bins=30, edgecolor='black')
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))
plt.plot(category_sales.index, category_sales.values, marker='o')
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
correlation = df[['Sales','Profit','Quantity','Discount']].corr()
correlation
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation,
            annot=True,
            cmap='coolwarm',
            linewidths=0.5)

plt.title("Correlation Heatmap")
plt.show()
```

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower)
print(upper)
```

```python
outliers = df[(df['Sales'] < lower) | (df['Sales'] > upper)]

print(outliers.shape)
outliers.head()
```

```python
df_clean = df[(df['Sales'] >= lower) & (df['Sales'] <= upper)]

print("Original Shape :", df.shape)
print("New Shape :", df_clean.shape)
```

The Sample Superstore dataset was analyzed to understand sales and profit performance across different regions, categories, and sub-categories.

The analysis provides useful business insights that can support decision-making and improve overall business performance.
