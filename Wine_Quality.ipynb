```python
corr = df.corr(numeric_only=True)
corr
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
corr["quality"].sort_values(ascending=False)
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
df[df["quality"] >= 7]
```

```python
df[df["alcohol"] > 12]
```

```python
df.sort_values(by="quality", ascending=False).head()
```

```python
df["quality"].value_counts().plot(kind="bar", color="skyblue")
plt.title("Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```

```python
df.groupby("quality")["alcohol"].mean().plot(marker="o")
plt.title("Average Alcohol Content by Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Average Alcohol")
plt.show()
```

```python
plt.hist(df["alcohol"], bins=20)
plt.title("Alcohol Distribution")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df["alcohol"], df["quality"])
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.title("Alcohol vs Quality")
plt.show()
```

## Observations

- Most wines have quality ratings between 5 and 6.
- Alcohol content generally increases with wine quality.
- Volatile acidity has a negative relationship with quality.
- Some numerical columns contain outliers.
- Correlation analysis shows alcohol is one of the strongest factors affecting wine quality.
