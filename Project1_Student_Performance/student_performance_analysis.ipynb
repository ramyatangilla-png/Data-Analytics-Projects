# EDA Summary - Student Performance Analysis

## Objective

The objective of this project is to analyze student exam performance and identify the factors that may affect students' scores in Mathematics, Reading, and Writing.

## Dataset Information

The Student Performance dataset contains information about students such as:

- Gender
- Race/Ethnicity
- Parental Level of Education
- Lunch Type
- Test Preparation Course
- Math Score
- Reading Score
- Writing Score

The dataset was analyzed using Python, NumPy, Pandas, and Matplotlib.

## Data Cleaning

The following data cleaning operations were performed:

- Checked the dataset for missing values.
- Checked for duplicate records.
- Removed duplicates if present.
- Examined numerical and categorical columns.
- Detected outliers using the IQR method.
- Treated outliers where necessary.

Missing values = 0
Duplicates = 0

## Statistical Analysis

Descriptive statistical measures were calculated for the numerical features:

- Mean
- Median
- Mode
- Variance
- Standard Deviation

These measures helped understand the central tendency and spread of student exam scores.

## Correlation Analysis

Correlation analysis was performed on Math, Reading, and Writing scores.

The analysis showed positive relationships between the three exam scores.

Reading and Writing scores showed a particularly strong relationship, indicating that students who perform well in reading also tend to perform well in writing.

|         | Math | Reading | Writing |
| ------- | ---: | ------: | ------: |
| Math    | 1.00 |    0.82 |    0.80 |
| Reading | 0.82 |    1.00 |    0.95 |
| Writing | 0.80 |    0.95 |    1.00 |


## Data Visualizations

Four types of visualizations were created during the analysis:

### 1. Bar Chart
The bar chart was used to compare the average Math, Reading, and Writing scores.

### 2. Line Chart
The line chart was used to observe variations in student scores across the dataset.

### 3. Histogram
The histogram was used to understand the distribution of Math scores.

### 4. Scatter Plot
The scatter plot was used to examine the relationship between Reading and Writing scores.

## Key Insights

- Student performance varies across Math, Reading, and Writing.
- Reading and Writing scores have a strong positive relationship.
- Students with higher Reading scores generally tend to have higher Writing scores.
- The histogram helps identify the most common range of student scores.
- Statistical analysis provides information about the average performance and variation in scores.
- Outlier analysis helps identify unusually high or low scores.

## Conclusion

The Student Performance Analysis project provided useful insights into students' exam performance.

Using Pandas and NumPy, the dataset was cleaned, explored, and statistically analyzed. Correlation analysis was used to study relationships between exam scores, while Matplotlib visualizations helped present the findings clearly.

The analysis indicates that student performance across Math, Reading, and Writing is related, with Reading and Writing showing a particularly strong relationship.
